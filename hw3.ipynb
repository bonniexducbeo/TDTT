```python
def gcd(a, b):
    r = a % b
    while( r != 0 ):
        a = b
        b = r
        r = a % b
    return b
a, b = map(int, input().split())

g = gcd(a, b)
for i in range(1, g + 1):
    if(g % i == 0):
        print(i, end = " ")
```

```python
#kiem tra so chinh phuong
def check(n):
    if n < 0:
        return False
    for i in range(1, n + 1):
        if i * i == n:
            return True
    return False
n = int(input())
if check(n):
    print("Yes")
else:
    print("No")
```

```python
#kiem tra so chinh phuong
def check(n):
    if n < 0:
        return False
    for i in range(n + 1):
        if i * i == n:
            return True
        if i * i > n:
            break
    return False
n = int(input())
if check(n):
    print("Yes")
else:
    print("No")
```

```python
#10
def tong_uoc(n):
    tong = 0
    for i in range(1, n):
        if( n % i == 0):
            tong += i
    return tong
a = int(input())
k = tong_uoc(a)
if(k == a):
    print("Yes")
else:
    print("No")
```
