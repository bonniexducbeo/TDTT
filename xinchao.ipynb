```python
n, m = map(int, input().split())
match n:
    case 4 | 6 | 9 | 11:
        print("30")
    case 2:
        if( m % 4 == 0 and m % 100 != 0 or m % 400 == 0):
            print("29")
        else:
            print("28")
    case 1 | 3 | 5 | 7 | 8 | 10 | 12:
        print("31")
    case _:
        print("ERROR")
```
